```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#-----Load the data  to a DataFrame
df = pd.read_csv("../data/students.csv")
df
```

```python
print(df.head() )   #------------5 FIRST ROWS
print(df.tail()  )  #---------5 last rows
df.info()          #------Info about the dataframe
df.describe()      #--------Statistics
```

```python
print(df['Math'])
print(df.loc[[0,1]])
```

```python
df.isnull()           #---Check for missing value
df.isnull().sum()   #---count the missing values
```

```python
#Create a new column  Average
df['Average'] = (df["Math"] +df["Physics"] + df["Chemistry"])/3
```

```python
df['Average']= df['Average'].round(2)
df
print(df.loc[df['Average'].idxmax()]) #Find the student with the highest average
print(df.loc[df['Average'].idxmin()]) #Find the student with the lowest average
print(df.sort_values(by='Average',ascending=True))
print(df[['Name','Average']])

df.to_csv("../reports/students_with_average.csv",index=False) #Create a new dataset
```

```python
#1 get students whose average is greater than 85?
gr85 = df['Average'] > 85
print(df[gr85])

#2 get only students whose Physics score is greater than 80?
ph80 = df['Physics'] > 80
print(df[ph80])

#3 get students whose Math score is less than 75?
mat75 = df['Math'] < 75
print(df[mat75])
```
