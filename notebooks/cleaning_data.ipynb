```python
import pandas as pd
df = pd.read_csv("../data/student_messy.csv")
```

```python
df.isnull()  #Finding missing values
df.isnull().sum() # Count missing values
df[df.isnull().any(axis=1)]
```

```python
#Fill missing value with average
df['Physics'] = df['Physics'].fillna(df['Physics'].mean())
df.to_csv("../reports/students_missing_cleaned.csv",index = False)
```

```python
#------------Finding duplicate Rows
df.duplicated()
print(df.duplicated().sum())
print(df[df.duplicated()])
df = df.drop_duplicates()
print(df.duplicated().sum())
df.to_csv("../reports/students_duplicates_cleaned.csv",index = False)
```

```python
#-------Fix Invalid Scores
df.loc[df['Math'] > 100, 'Math'] =  100
#--------Recalculate average
df['Average'] = (df['Math'] + df['Physics'] + df['Chemistry'])/3
df['Average'] = df['Average'].round(2) 
```
