```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("../reports/students_with_average.csv")
```

```python
#--------Average Grades
plt.figure(figsize=(8,5))   #Create an empty canvas width,height =8,5
plt.bar( df['Name'],df['Average'])
plt.title("Student Average Grades")
plt.xlabel('Students')
plt.ylabel('Average')
plt.xticks(rotation=30)  # rotates the names by 30 degrees
#plt.savefig('../reports/average_grades.png')  save figure
```

```python
#-----------Math Scores
plt.figure(figsize=(8,5))
plt.bar( df['Name'], df['Math'])
plt.title('Math Scores')
plt.xlabel('Students')
plt.ylabel('Score')
```

```python
#------Average Distribution
plt.figure(figsize=(8,5))
plt.hist(df['Average'])
plt.title("Average Distribution")
plt.xlabel('Average')
plt.ylabel('Number of Students')
plt.show()
```

```python
#-------Compare Subjects
subjects= ['Math', 'Physics','Chemistry']

means = [df['Math'].mean(),df['Physics'].mean(),df['Chemistry'].mean()]

plt.figure(figsize=(8,5))
plt.bar(subjects,means)
plt.title('Average Score per Subject')
plt.ylabel('Average Score')

    
```

```python
sara = df[df['Name'] == 'Sara']
print(sara)
scores = [sara['Math'].loc[1], sara['Physics'].loc[1], sara['Chemistry'].loc[1]]
print(scores)
plt.figure(figsize=(8,5))
plt.bar(subjects,scores)
plt.title('Sara Grades')
plt.ylabel('Scores')
plt.savefig("../reports/sara_grades.png")
```
